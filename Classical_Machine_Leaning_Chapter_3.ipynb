```python
import numpy as np

# Patient Data (Age, BMI,Blood Glucose Level , Actual Class)
data = np.array([
    [45, 29.5, 130, 1],
    [34, 22.0, 95, 0],
    [50, 31.2, 145, 1],
    [28, 19.8, 85, 0],
    [60, 33.1, 160, 1],
    [41, 25.0, 110, 0],
    [55, 30.5, 135, 1],
    [38, 27.4, 102, 0],
    [48, 28.7, 140, 1],
    [29, 23.5, 89, 0]
])

# Features and Labels
X = data[:, :-1]  # Features (Age, BMI, Blood Glucose Level)
y = data[:, -1]   # Labels (0 or 1 for diabetes)

# Add intercept column of 1s
intercept = np.ones((X.shape[0], 1))
X = np.hstack((intercept, X))

# Parameters
learning_rate = 0.01  # Reduced learning rate
num_iterations = 10000  # Lower number of iterations

# Initialize weights (including intercept)
weights = np.zeros(X.shape[1])
loss = 0
# Sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Training the model
for i in range(num_iterations):
    # Linear model
    z = np.dot(X, weights)
    # Sigmoid activation
    predictions = sigmoid(z)

    # Compute gradients
    dw = np.dot(X.T, (predictions - y)) / y.size

    # Update weights
    weights -= learning_rate * dw

    # Optional: print loss every 1000 iterations
    if i % 1000 == 0:
        # Loss calculation with epsilon to avoid log(0)
        epsilon = 1e-10
        loss = -np.mean(y * np.log(predictions + epsilon) + (1 - y) * np.log(1 - predictions + epsilon))
        print(f"Iteration {i}: Loss = {loss}")
print(loss)
# Final weights including intercept
print("Final Weights (including intercept):", weights)

# Final predictions (0 or 1)
final_probabilities = sigmoid(np.dot(X, weights))
final_predictions = (final_probabilities >= 0.5).astype(int)

print("Final Predictions (0 or 1):", final_predictions)
```
