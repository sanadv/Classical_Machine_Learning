```python
import tensorflow as tf
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from tensorflow.keras.models import Sequential

# Example model to demonstrate feature extraction using multiple Conv2D layers
model = Sequential([
    # First Convolutional Layer
    Conv2D(filters=32, kernel_size=(3, 3),padding='same', activation='relu', input_shape=(64, 64, 3)),

    # Second Convolutional Layer
    Conv2D(filters=64, kernel_size=(3, 3) ,padding='same', activation='relu'),

    # Third Convolutional Layer
    Conv2D(filters=128, kernel_size=(3, 3) ,padding='same', activation='relu'),

    # Flatten the feature maps
    Flatten(),

    # Fully connected ANN layers
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')  # Assuming we have 10 output classes
])

# Display the model architecture
model.summary()
```

```python
import tensorflow as tf
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from tensorflow.keras.models import Sequential

# Example model to demonstrate feature extraction using multiple Conv2D layers
model = Sequential([
    # First Convolutional Layer
    Conv2D(filters=32, kernel_size=(3, 3),padding='same', activation='relu', input_shape=(64, 64, 3)),
    MaxPooling2D(pool_size=(2, 2)),
    # Second Convolutional Layer
    Conv2D(filters=64, kernel_size=(3, 3) ,padding='same', activation='relu'),
    MaxPooling2D(pool_size=(2, 2)),
    # Third Convolutional Layer
    Conv2D(filters=128, kernel_size=(3, 3) ,padding='same', activation='relu'),
    MaxPooling2D(pool_size=(2, 2)),
    # Flatten the feature maps
    Flatten(),

    # Fully connected ANN layers
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')  # Assuming we have 10 output classes
])

# Display the model architecture
model.summary()
```
